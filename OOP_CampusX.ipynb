### 1. Everything about class:

```python
class Atm:
  def __init__(self):
    self.pin = '1234'
    self.balance = 120
    self.menu()

  def menu(self):
      user_input = input("""
    Hello, how would you like to proceed ?
    1. Enter 1 to create pin.
    2. Enter 2 to deposit
    3. Entr 3 to withdraw.
    4. Enter 4 to check balance.
    5. Enter 5 to exit.
    """)

      if user_input == '1':
          self.create_pin()
      elif user_input == '2':
          self.deposit()
      elif user_input == '3':
          self.withdraw()
      elif user_input == '4':
          self.check_balance()
      else:
          print("bye")

  def create_pin(self):
    self.pin = input("Enter the pin: ")
    print("Pin set successfully")

  def deposit(self):
    temp = input("Enter your pin: ")
    if temp == self.pin:
      amount = int(input("Enter the amount to deposit: "))
      self.balance = self.balance + amount
      print(f"Deposit successfully! Your updated balance is: {self.balance}")
    else:
      print("Invalid pin!")
      self.menu()

  def withdraw(self):
    temp = input("Enter your pin: ")
    if temp == self.pin:
      amount = int(input("Enter the amount to withdraw: "))
      if amount < self.balance:
        self.balance = self.balance - amount
        print(f"Withdraw successfully! Your updated balance is: {self.balance}")
      else:
        print(f"Insufficient amount! Add {abs(self.balance- amount)} to withdraw.")
    else:
      print("Invalid pin!")
      self.menu()

  def check_balance(self):
    temp = input("Enter your pin: ")
    if temp == self.pin:
      print(f"Your balance is: {self.balance}")
      print("Thanks!")
```

```python
## Ops, there is no fraction details when we are printing the variable, lets use magic function here:

class Fraction:
  def __init__(self,n,d):
    self.num = n
    self.den = d

  def __str__(self):
    return f"{self.num}/{self.den}"

  def __add__(self,other):
    temp_num = self.num * other.den + self.den * other.num
    temp_den = self.den * self.num
    return f"{temp_num}/{temp_den}"

  def __sub__(self,other):
    temp_num = self.num * other.den - self.den * other.num
    temp_den = self.den * self.num
    return f"{temp_num}/{temp_den}"

  def __mul__(self,other):
    temp_num = self.num * other.num
    temp_den = self.den * other.den
    return f"{temp_num}/{temp_den}"

  def __truediv__(self,other):
    temp_num = self.num * other.den
    temp_den = self.den * other.num
    return f"{temp_num}/{temp_den}"
```

```python
## Add
x = Fraction(3,4)
y = Fraction(4,5)
print(x)
print(y)

print(x+y)
```

```python
## Sub
x = Fraction(3,4)
y = Fraction(4,5)
print(x)
print(y)

print(x-y)
```

```python
## Mul
x = Fraction(3,4)
y = Fraction(4,5)
print(x)
print(y)

print(x*y)
```

```python
## Div
x = Fraction(3,4)
y = Fraction(4,5)
print(x)
print(y)

print(x/y)
```

```python
## Lets try to understand get_pin and set_pin.
class Atm:
  def __init__(self):
    self.__pin = '1234'
    self.__balance = 120
    self.menu()

  def get_pin(self):
    return self.__pin

  def set_pin(self, new_pin):
    if new_pin == str:
      self.__pin = new_pin
      print("Pin changed")
    else:
      print("Not allowed")

  def menu(self):
      user_input = input("""
    Hello, how would you like to proceed ?
    1. Enter 1 to create pin.
    2. Enter 2 to deposit
    3. Entr 3 to withdraw.
    4. Enter 4 to check balance.
    5. Enter 5 to exit.
    """)

      if user_input == '1':
          self.create_pin()
      elif user_input == '2':
          self.deposit()
      elif user_input == '3':
          self.withdraw()
      elif user_input == '4':
          self.check_balance()
      else:
          print("bye")

  def create_pin(self):
    old_pin_confirmation = input("Enter your old pin: ")
    if self.__pin == old_pin_confirmation:
      self.__pin = input("Enter your new pin: ")
      print("Pin set successfully")
    else:
      print("Old pin did not match! Try again.")

  def deposit(self):
    temp = input("Enter your pin: ")
    if temp == self.__pin:
      amount = int(input("Enter the amount to deposit: "))
      self.__balance = self.__balance + amount
      print(f"Deposit successfully! Your updated balance is: {self.__balance}")
    else:
      print("Invalid pin!")
      self.menu()

  def withdraw(self):
    temp = input("Enter your pin: ")
    if temp == self.__pin:
      amount = int(input("Enter the amount to withdraw: "))
      if amount < self.__balance:
        self.__balance = self.__balance - amount
        print(f"Withdraw successfully! Your updated balance is: {self.__balance}")
      else:
        print(f"Insufficient amount! Add {abs(self.__balance- amount)} to withdraw.")
    else:
      print("Invalid pin!")
      self.menu()

  def check_balance(self):
    temp = input("Enter your pin: ")
    if temp == self.__pin:
      print(f"Your balance is: {self.__balance}")
      print("Thanks!")
    else:
      print("Invalid pin!")
```
