#### Aggregation

```python
# Aggregation (Has - A)
'''
Customer ----> has A ----> Address
'''
```

```python
class Customer:

  def __init__(self, name, gender, address):
    self.name = name
    self.gender = gender
    self.address = address

class Address:

  def __init__(self, city, pincode, state):
    self.city = city
    self.pincode = pincode
    self.state = state

  def change_address(self, new_city, new_pincode, new_state):
    self.city = new_city
    self.pincode = new_pincode
    self.state = new_state
```

```python
add = Address('Lucknow', '226001', 'Uttar Pradesh')
cust = Customer('Shahrukh', 'Male', add)
```

```python
print(cust.address.city)
print(cust.address.pincode)
print(cust.address.state)
```

```python
## Now changing address
# BEFORE
add = Address('Lucknow', '226001', 'Uttar Pradesh')
cust = Customer('Shahrukh', 'Male', add)
print(cust.address.city)
print(cust.address.pincode)
print(cust.address.state)

# AFTER
add.change_address(new_city = 'Bangalore', new_pincode = 123456, new_state = 'Karnataka')
cust = Customer('Shahrukh', 'Male', add)
print(cust.address.city)
print(cust.address.pincode)
print(cust.address.state)
```

```python
## Now lets add one more methon

class Customer:

  def __init__(self, name, gender, address):
    self.name = name
    self.gender = gender
    self.address = address

  def edit_profile(self, new_name, new_city, new_pincode, new_state):
    self.name = new_name
    self.address.change_address(new_city, new_pincode, new_state)

class Address:

  def __init__(self, city, pincode, state):
    self.city = city
    self.pincode = pincode
    self.state = state

  def change_address(self, new_city, new_pincode, new_state):
    self.city = new_city
    self.pincode = new_pincode
    self.state = new_state
```

```python
add = Address('Lucknow', '226001', 'Uttar Pradesh')
cust = Customer('Shahrukh', 'Male', add)

cust.edit_profile("Ankit", "Gurgaon", 122001, "Haryana")

print(cust.address.pincode)
```

#### Inheritance

```python
# NOTE: Private members (__) can not be Inherited.
```

```python
class User:

  def login(self):
    print("Login")

  def register(self):
    print("Register")

class Student(User):

  def enroll(self):
    print("Enroll")

  def review(self):
    print("Review")
```

```python
st1 = Student()
st1.enroll(), st1.review(), st1.login(), st1.register()
```
