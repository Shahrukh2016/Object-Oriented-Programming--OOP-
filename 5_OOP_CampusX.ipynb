```python
## Example of multi level inheritance:

class Product:
  def review(self):
    print ("Product customer review")

class Phone(Product):
  def __init__ (self, price, brand, camera):
    print ("Inside phone constructor")
    self. price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print ("Buying a phone")

class SmartPhone(Phone):
  pass
```

```python
s = SmartPhone(20000, "Apple", 12)
p = Phone(1000,"Samsung", 1)
s.buy
s.review()
p.review()
```

```python
##### Hierarchical Inheritance #######################
class Phone:
  def __init__ (self, price, brand, camera):
    print ("Inside phone constructor")
    self. price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print ("Buying a phone")

  def return_phone(self):
    print ("Returning a phone")

class SmartPhone(Phone):
  pass

class FeaturePhone(Phone):
  pass

SmartPhone(1000,"Apple","13px").buy()
```

```python
############################ Multiple Inheritance #######################

class Phone:
  def __init__(self, price, brand, camera):
    print ("Inside phone constructor")
    self.price = price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print ("Buying a phone")

class Product:
  def review(self):
    print ("Customer review")

class SmartPhone(Phone, Product):
  pass
```

```python
s = SmartPhone(20000, "Apple", 12)
s.buy()
s.review()
```

```python
class Phone:
  def __init__(self, price, brand, camera):
    print ("Inside phone constructor")
    self. price =  price
    self.brand = brand
    self.camera = camera

  def buy(self):
    print ("Buying a phone")

class Product:
  def buy(self):
    print ("Product buy method")

class SmartPhone (Phone, Product):
  pass
```

```python
s = SmartPhone(20000, "Apple", 12)
s.buy()

## Buy method of Phone class will be called, as it is the first inheritance than Product. This process is also known as MRO (Method Resolution Order).
```

```python
### Example of Multilevel Inheritance

class A:
  def m1 (self):
    return 20

class B(A):
  def m1 (self):
    return 30

  def m2(self):
    return 40

class C(B):
  def m2(self):
    return 20
```

```python
obj1 = A()
obj2 = B()
obj3 = C()
print(obj1.m1() + obj3.m1()+ obj3.m2())
```

```python
class A:
  def m1(self):
    return 20

class B(A):
  def m1(self):
    val = super().m1()+30
    return val

class C(B):
  def m1(self):
    val=self.m1()+20
    return val
```

```python
obj=C()
print(obj.m1())
```
