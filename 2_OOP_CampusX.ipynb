```python
class Customer:

  def __init__(self,name):
    self.name = name
```

```python
class Customer:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

def greet(customer):
  if customer.gender == "Male":
    print("Hello", customer.name, "sir")
  else:
    print("Hello", customer.name, "ma'am")
```

```python
cust = Customer("Shahrukh", "Male")

greet(cust)
```

```python
class Customer:

  def __init__(self,name,gender):
    self.name = name
    self.gender = gender

def greet(customer):
  if customer.gender == "Male":
    print("Hello", customer.name, "sir")
  else:
    print("Hello", customer.name, "ma'am")

  cust2 = Customer("Sakshi", "Female")
  return cust2.name, cust2.gender
```

```python
cust = Customer("Shahrukh", "Male")

greet(cust)
```

```python
'''
From above codes we can say that we can do 2 things with classes:
1. We can call our defined object in any function. ex: greet(cust)
2. A defined function can return the created object. ex: cust2
'''
```

```python
### Now comes what is 'Pass by reference'
```

```python
class Customer:

  def __init__(self,name):
    self.name = name

def greet(customer):
  print(id(customer))
```

```python
cust = Customer("Shahrukh")
print(id(cust))
```

```python
greet(cust)
```

```python
## In above code since the reference is same, it is also pointing the same id and space.
```

```python
class Customer:

  def __init__(self,name):
    self.name = name

def greet(customer):
  # print(id(customer))
  customer.name = "Sakshi"
  print(f"greet print: {customer.name}")
```

```python
cust = Customer("Shahrukh")
greet(cust)

print(f"Class print: {cust.name}")
```

```python
#### From the above code we have observed if we are overwriting the class data (object) inside function, then it will also change the original data..

## That means class objects are also MUTABLE in nature like lists, dicts.
```
